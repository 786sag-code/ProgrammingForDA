```python

import pandas as pd

# 1. Create a DataFrame with student names and scores
data = {
    "Name": ["Mary", "Jack", "Ahmed", "June", "Hassan"],
    "Score": [36, 73, 56, 66, 87]
}

df = pd.DataFrame(data)

print("Original Student DataFrame:")
print(df)

# 2. Add a Rank column (highest score gets rank 1)
df["Rank"] = df["Score"].rank(
    ascending=False,
    method="min"
).astype(int)

print("\nDataFrame with Ranks:")
print(df)

# 3. Sort by Score in descending order
sorted_df = df.sort_values(
    by="Score",
    ascending=False
)

print("\nStudents Sorted by Score:")
print(sorted_df[["Name", "Score", "Rank"]])

# 4. Filter students with a score of at least 60
top_students = sorted_df[sorted_df["Score"] >= 60]

print("\nStudents Scoring 60 or Above:")
print(top_students[["Name", "Score", "Rank"]])
```

```python

import pandas as pd
import numpy as np

# Create a sample DataFrame with missing values
df = pd.DataFrame({
    "Name": ["Mary", "Jack", "Ahmed", "June", "Hassan"],
    "Score": [36, 73, np.nan, 66, 87],
    "Age": [22, np.nan, 25, 23, 24]
})

# Check the original data
print("Original DataFrame:")
print(df)

print("\nMissing values before cleaning:")
print(df.isna().sum())

# Clean the data by removing rows with missing values
df_clean = df.dropna()

# Check the cleaned data
print("\nCleaned DataFrame:")
print(df_clean)

print("\nMissing values after cleaning:")
print(df_clean.isna().sum())

# Compare the number of rows before and after cleaning
print("\nRows before cleaning:", len(df))
print("Rows after cleaning:", len(df_clean))
```

```python

import numpy as np

# Example 1: Typical data
scores = np.array([5, 6, 7, 8, 9], dtype=float)

print("=== Typical Data ===")
print("Scores:", scores)
print("Mean:", np.mean(scores))
print("Median:", np.median(scores))

# Z-score formula: z = (x - mean) / standard deviation
mean = np.mean(scores)
std = np.std(scores)  # Population standard deviation
z_scores = (scores - mean) / std

print("Standard deviation:", round(std, 2))
print("Z-scores:", np.round(z_scores, 2))
print("Potential outliers:", scores[np.abs(z_scores) > 3])

print("\n=== Data With an Outlier ===")

# Example 2: Add an extreme value
scores_outlier = np.array([5, 6, 7, 8, 100], dtype=float)

print("Scores:", scores_outlier)
print("Mean:", np.mean(scores_outlier))
print("Median:", np.median(scores_outlier))

mean2 = np.mean(scores_outlier)
std2 = np.std(scores_outlier)
z_scores2 = (scores_outlier - mean2) / std2

print("Standard deviation:", round(std2, 2))
print("Z-scores:", np.round(z_scores2, 2))
print("Potential outliers:",
      scores_outlier[np.abs(z_scores2) > 3])
```

```python

import numpy as np

# Student heights (cm)
height = np.array([148, 154, 105, 108, 111, 116, 133, 149])

# 1. Minimum, maximum, and range
print("Minimum:", np.min(height))
print("Maximum:", np.max(height))
print("Range:", np.ptp(height))

# 2. 90th percentile
print("90th percentile:", np.percentile(height, 90))

# 3. Mean, median, and average
print("Mean:", np.mean(height))
print("Median:", np.median(height))
print("Average:", np.average(height))

# 4. Variance and standard deviation
print("Variance:", np.var(height))
print("Standard deviation:", np.std(height))

# 5. Interpretation
print("\nInterpretation:")
print("Heights are measured in centimeters.")
print("Range and standard deviation show the spread of heights.")
print("Variance is measured in square centimeters (cm²).")
print("The 90th percentile is a height threshold, not the maximum.")
```

```python

import numpy as np

# Define matrices A and B
A = np.array([
    [8, 4, 1],
    [2, -4, 6],
    [5, 3, -1]
])

B = np.array([
    [6, 7, -5],
    [3, -1, -1],
    [4, 8, -2]
])

# Define matrix H
H = np.array([
    [2, 3, -3, -3],
    [-8, -1, -3, 4],
    [-6, 6, 0, -2],
    [1, 2, 9, 7]
], dtype=float)

# Task 1: Matrix arithmetic
print("A + B:\n", A + B)
print("\nA - B:\n", A - B)
print("\n3 * A:\n", 3 * A)
print("\nB - A:\n", B - A)

# Task 2: Define C and F
C = A + B
F = C / 4

print("\nC = A + B:\n", C)
print("\nF = C / 4:\n", F)
print("\nB - 2F:\n", B - 2 * F)

# Task 3: Element-wise vs matrix multiplication
print("\nA * B (element-wise):\n", A * B)
print("\nA @ B (matrix multiplication):\n", A @ B)

# Task 4: Find and verify inverse of H
H_inv = np.linalg.inv(H)

print("\nDeterminant of H:", np.linalg.det(H))
print("\nInverse of H:\n", H_inv)
print("\nH @ H_inv:\n", H @ H_inv)
print("\nInverse verified:", np.allclose(H @ H_inv, np.eye(4)))
```

```python
import numpy as np

# 1. Create product names and revenue arrays
products = np.array(["Laptop", "Phone", "Tablet", "Headphones"])
revenues = np.array([2500, 4200, 1800, 3200])

print("Products:")
print(products)

print("\nRevenues:")
print(revenues)


# 2. Sort revenue ascending and descending

ascending = np.sort(revenues)
descending = np.sort(revenues)[::-1]

print("\nRevenue ascending:")
print(ascending)

print("\nRevenue descending:")
print(descending)


# 3. Use argsort() to rank products by revenue

# argsort gives the INDEX positions that would sort the revenues
ascending_index = np.argsort(revenues)

# Reverse the indexes for highest revenue first
descending_index = np.argsort(revenues)[::-1]

print("\nProduct ranking - highest to lowest:")

for index in descending_index:
    print(products[index], "€", revenues[index])


# 4. Report the top two products and their revenues

top_two = descending_index[:2]

print("\nTop two products:")

for index in top_two:
    print(products[index], "€", revenues[index])
```
