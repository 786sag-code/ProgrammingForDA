```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy.interpolate import PchipInterpolator, CubicSpline
from scipy.cluster.hierarchy import linkage, fcluster

from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv('../data/mona.csv')

# Identify date column
date_col = next(
    col for col in df.columns
    if col.lower() == 'date'
)

df[date_col] = pd.to_datetime(df[date_col])

df = df.set_index(date_col).sort_index()

# Identify CO2 measurement column
co2_col = df.select_dtypes(include='number').columns[0]

co2 = df[co2_col].astype(float)

# Handle duplicate dates, if any
co2 = co2.groupby(level=0).mean()

print("Dataset shape:", df.shape)
print("CO2 column:", co2_col)
print("Missing observations:", co2.isna().sum())
print("Date range:", co2.index.min(), "to", co2.index.max())

display(co2.head())
```

```python
# Keep only observed values
observed = co2.dropna()

# Reproducible random selection
np.random.seed(42)

# Hide 10% of known readings
holdout_size = int(len(observed) * 0.10)

holdout_dates = np.random.choice(
    observed.index,
    size=holdout_size,
    replace=False
)

# Actual values for evaluation
actual = observed.loc[holdout_dates]

# Training data with hidden values removed
training = co2.copy()
training.loc[holdout_dates] = np.nan

print("Original observations:", len(observed))
print("Hidden observations:", len(holdout_dates))
print("Training observations:", training.notna().sum())
```

```python
def interpolate_co2(series, method):

    known = series.dropna()

    # Convert dates into numerical values
    x = (
        known.index - series.index[0]
    ).total_seconds() / (24 * 60 * 60)

    x_all = (
        series.index - series.index[0]
    ).total_seconds() / (24 * 60 * 60)

    y = known.values

    if method == 'linear':
        interpolator = np.interp(
            x_all, x, y,
            left=np.nan,
            right=np.nan
        )

    elif method == 'pchip':
        model = PchipInterpolator(x, y)
        interpolator = model(x_all)

    elif method == 'cubic':
        model = CubicSpline(x, y)
        interpolator = model(x_all)

    result = series.copy()

    missing = result.isna()

    result.loc[missing] = interpolator[missing]

    return result
```

```python
linear_pred = interpolate_co2(training, 'linear')

pchip_pred = interpolate_co2(training, 'pchip')

cubic_pred = interpolate_co2(training, 'cubic')

print("All three interpolation methods completed.")
```

```python
from sklearn.metrics import mean_squared_error

predictions = {
    'Linear': linear_pred.loc[holdout_dates],
    'PCHIP': pchip_pred.loc[holdout_dates],
    'Cubic Spline': cubic_pred.loc[holdout_dates]
}

results = []

for method, predicted in predictions.items():

    rmse = np.sqrt(
        mean_squared_error(actual, predicted)
    )

    results.append({
        'Method': method,
        'RMSE': rmse
    })

comparison = pd.DataFrame(results)

print(comparison.sort_values('RMSE'))

best_method = comparison.loc[
    comparison['RMSE'].idxmin(), 'Method'
]

print("\nBest interpolation method:", best_method)
```

```python
missing_positions = np.flatnonzero(co2.isna().values)

if len(missing_positions) == 0:
    print("No original missing gaps found.")
else:
    # Split missing positions into consecutive groups
    groups = np.split(
        missing_positions,
        np.where(np.diff(missing_positions) != 1)[0] + 1
    )

    # Select longest missing period
    longest_gap = max(groups, key=len)

    gap_start = co2.index[longest_gap[0]]
    gap_end = co2.index[longest_gap[-1]]

    print("Gap starts:", gap_start)
    print("Gap ends:", gap_end)
    print("Missing weeks:", len(longest_gap))
```

```python
# Interpolate original dataset
linear_full = interpolate_co2(co2, 'linear')
pchip_full = interpolate_co2(co2, 'pchip')
cubic_full = interpolate_co2(co2, 'cubic')

if len(missing_positions) > 0:

    # Select a window around the longest gap
    start = gap_start - pd.Timedelta(weeks=8)
    end = gap_end + pd.Timedelta(weeks=8)

    plt.figure(figsize=(13, 6))

    plt.plot(
        co2.loc[start:end],
        'o',
        label='Original observations',
        markersize=3
    )

    plt.plot(
        linear_full.loc[start:end],
        label='Linear'
    )

    plt.plot(
        pchip_full.loc[start:end],
        label='PCHIP'
    )

    plt.plot(
        cubic_full.loc[start:end],
        label='Cubic spline'
    )

    plt.title('Comparison of Interpolation Methods')
    plt.xlabel('Date')
    plt.ylabel('CO2 Concentration (ppm)')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
method_map = {
    'Linear': linear_full,
    'PCHIP': pchip_full,
    'Cubic Spline': cubic_full
}

co2_filled = method_map[best_method]

print("Selected method:", best_method)
print("Remaining missing values:", co2_filled.isna().sum())
```

```python
monthly = co2_filled.resample('MS').mean()

# Create Year x Month table
monthly_table = monthly.to_frame('CO2')

monthly_table['Year'] = monthly_table.index.year
monthly_table['Month'] = monthly_table.index.month

year_month = monthly_table.pivot(
    index='Year',
    columns='Month',
    values='CO2'
)

print(year_month)
```

```python
# Remove years without all 12 monthly observations
year_month = year_month.dropna()

# Centre each year by subtracting its annual monthly mean
centred_profiles = year_month.sub(
    year_month.mean(axis=1),
    axis=0
)

print(centred_profiles.head())
```

```python
# Number of clusters
number_of_clusters = 3

# Hierarchical clustering
Z = linkage(
    centred_profiles.values,
    method='ward'
)

# Assign cluster labels
cluster_labels = fcluster(
    Z,
    t=number_of_clusters,
    criterion='maxclust'
)

# Add cluster labels to the table
centred_profiles['Cluster'] = cluster_labels

print(centred_profiles[['Cluster']])
```

```python
plt.figure(figsize=(12, 6))

months = range(1, 13)

for cluster in sorted(centred_profiles['Cluster'].unique()):

    cluster_data = centred_profiles[
        centred_profiles['Cluster'] == cluster
    ].drop(columns='Cluster')

    # Plot average seasonal profile for each cluster
    average_profile = cluster_data.mean(axis=0)

    plt.plot(
        months,
        average_profile,
        marker='o',
        label=f'Cluster {cluster}'
    )

plt.title('Seasonal CO2 Profiles by Cluster')
plt.xlabel('Month')
plt.ylabel('CO2 Deviation from Annual Mean (ppm)')

plt.xticks(
    months,
    ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
     'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
)

plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import mean_squared_error

# Load dataset
df = pd.read_csv("../data/mona.csv")

print(df.head())
print(df.info())
print(df.isnull().sum())
```

```python
# Convert date into datetime
df['date'] = pd.to_datetime(df['date'])

# Convert CO2 values into numeric
df['ppm'] = pd.to_numeric(df['co2_ppm'], errors='coerce')

# Remove missing observations
df = df.dropna(subset=['date', 'ppm'])

# Sort observations by date
df = df.sort_values('date').reset_index(drop=True)

# Calculate elapsed years
df['elapsed_years'] = (
    (df['date'] - df['date'].min()).dt.days / 365.2425
)

print(df.head())
print("Total observations:", len(df))
```

```python
# Annual seasonal angle
angle = 2 * np.pi * (
    df['date'].dt.dayofyear / 365.2425
)

# Seasonal features
df['sin_season'] = np.sin(angle)
df['cos_season'] = np.cos(angle)

# Target variable
y = df['ppm'].values

# Model A: Trend only
X_trend = np.column_stack([
    np.ones(len(df)),
    df['elapsed_years']
])

# Model B: Trend + seasonality
X_seasonal = np.column_stack([
    np.ones(len(df)),
    df['elapsed_years'],
    df['sin_season'],
    df['cos_season']
])

print("Trend matrix:", X_trend.shape)
print("Seasonal matrix:", X_seasonal.shape)
```

```python
# Fit trend-only model
beta_trend = np.linalg.lstsq(
    X_trend, y, rcond=None
)[0]

# Fit trend + seasonal model
beta_seasonal = np.linalg.lstsq(
    X_seasonal, y, rcond=None
)[0]

# Predictions
pred_trend = X_trend @ beta_trend
pred_seasonal = X_seasonal @ beta_seasonal

# Calculate RMSE
rmse_trend = np.sqrt(mean_squared_error(y, pred_trend))

rmse_seasonal = np.sqrt(
    mean_squared_error(y, pred_seasonal)
)

print("Trend-only RMSE:", round(rmse_trend, 2))
print("Seasonal model RMSE:", round(rmse_seasonal, 2))

print("Trend slope:", round(beta_seasonal[1], 2), "ppm/year")
```

```python
coefficients = pd.DataFrame({
    'Feature': [
        'Intercept',
        'Elapsed Years',
        'Sine Season',
        'Cosine Season'
    ],
    'Coefficient': beta_seasonal
})

print(coefficients)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df['date'], y, label='Actual CO2')
plt.plot(df['date'], pred_trend, label='Trend only')
plt.plot(df['date'], pred_seasonal, label='Trend + Seasonality')

plt.xlabel('Year')
plt.ylabel('CO2 concentration (ppm)')
plt.title('Mauna Loa CO2: Regression Models')
plt.legend()
plt.grid(True)

plt.show()
```

```python
residuals = y - pred_seasonal

plt.figure(figsize=(12, 5))

plt.scatter(df['date'], residuals, s=8)
plt.axhline(y=0, linestyle='--')

plt.xlabel('Year')
plt.ylabel('Residuals (ppm)')
plt.title('Residual Plot – Seasonal Model')
plt.grid(True)

plt.show()
```

```python
# Add squared elapsed years
X_quadratic = np.column_stack([
    np.ones(len(df)),
    df['elapsed_years'],
    df['elapsed_years'] ** 2,
    df['sin_season'],
    df['cos_season']
])

# Fit quadratic model
beta_quad = np.linalg.lstsq(
    X_quadratic, y, rcond=None
)[0]

pred_quad = X_quadratic @ beta_quad

rmse_quad = np.sqrt(
    mean_squared_error(y, pred_quad)
)

print("Quadratic model RMSE:", round(rmse_quad, 2))
```

```python
# Calculate anomaly above 315 ppm
anomaly = (co2_clean - 320).clip(lower=0)

# Convert dates into elapsed weeks
time_weeks = (
    (co2_clean.index - co2_clean.index[0])
    .total_seconds() / (7 * 24 * 3600)
)

# Calculate area using trapezoidal rule
total_area = np.trapezoid(
    anomaly.values,
    x=time_weeks
)

print("Total accumulated CO2 anomaly:", total_area)
print("Unit: ppm-weeks")
```

```python
# Calculate area between consecutive observations
dt = np.diff(time_weeks)

areas = (
    (anomaly.values[:-1] + anomaly.values[1:])
    / 2
) * dt

# Calculate cumulative area
cumulative_area = np.concatenate([
    [0],
    np.cumsum(areas)
])

# Plot
plt.figure(figsize=(12, 6))

plt.plot(
    co2_clean.index,
    cumulative_area
)

plt.title("Cumulative CO2 Anomaly Above 320 ppm")
plt.xlabel("Year")
plt.ylabel("Cumulative Area (ppm-weeks)")
plt.grid(True)

plt.show()
```

```python

import numpy as np

# Example 1: Typical data
scores = np.array([5, 6, 7, 8, 9], dtype=float)

print("=== Typical Data ===")
print("Scores:", scores)
print("Mean:", np.mean(scores))
print("Median:", np.median(scores))

# Z-score formula: z = (x - mean) / standard deviation
mean = np.mean(scores)
std = np.std(scores)  # Population standard deviation
z_scores = (scores - mean) / std

print("Standard deviation:", round(std, 2))
print("Z-scores:", np.round(z_scores, 2))
print("Potential outliers:", scores[np.abs(z_scores) > 3])

print("\n=== Data With an Outlier ===")

# Example 2: Add an extreme value
scores_outlier = np.array([5, 6, 7, 8, 100], dtype=float)

print("Scores:", scores_outlier)
print("Mean:", np.mean(scores_outlier))
print("Median:", np.median(scores_outlier))

mean2 = np.mean(scores_outlier)
std2 = np.std(scores_outlier)
z_scores2 = (scores_outlier - mean2) / std2

print("Standard deviation:", round(std2, 2))
print("Z-scores:", np.round(z_scores2, 2))
print("Potential outliers:",
      scores_outlier[np.abs(z_scores2) > 3])
```

```python

import numpy as np

# Define matrices A and B
A = np.array([
    [8, 4, 1],
    [2, -4, 6],
    [5, 3, -1]
])

B = np.array([
    [6, 7, -5],
    [3, -1, -1],
    [4, 8, -2]
])

# Define matrix H
H = np.array([
    [2, 3, -3, -3],
    [-8, -1, -3, 4],
    [-6, 6, 0, -2],
    [1, 2, 9, 7]
], dtype=float)

# Task 1: Matrix arithmetic
print("A + B:\n", A + B)
print("\nA - B:\n", A - B)
print("\n3 * A:\n", 3 * A)
print("\nB - A:\n", B - A)

# Task 2: Define C and F
C = A + B
F = C / 4

print("\nC = A + B:\n", C)
print("\nF = C / 4:\n", F)
print("\nB - 2F:\n", B - 2 * F)

# Task 3: Element-wise vs matrix multiplication
print("\nA * B (element-wise):\n", A * B)
print("\nA @ B (matrix multiplication):\n", A @ B)

# Task 4: Find and verify inverse of H
H_inv = np.linalg.inv(H)

print("\nDeterminant of H:", np.linalg.det(H))
print("\nInverse of H:\n", H_inv)
print("\nH @ H_inv:\n", H @ H_inv)
print("\nInverse verified:", np.allclose(H @ H_inv, np.eye(4)))
```

```python
import numpy as np

# 1. Create product names and revenue arrays
products = np.array(["Laptop", "Phone", "Tablet", "Headphones"])
revenues = np.array([2500, 4200, 1800, 3200])

print("Products:")
print(products)

print("\nRevenues:")
print(revenues)


# 2. Sort revenue ascending and descending

ascending = np.sort(revenues)
descending = np.sort(revenues)[::-1]

print("\nRevenue ascending:")
print(ascending)

print("\nRevenue descending:")
print(descending)


# 3. Use argsort() to rank products by revenue

# argsort gives the INDEX positions that would sort the revenues
ascending_index = np.argsort(revenues)

# Reverse the indexes for highest revenue first
descending_index = np.argsort(revenues)[::-1]

print("\nProduct ranking - highest to lowest:")

for index in descending_index:
    print(products[index], "€", revenues[index])


# 4. Report the top two products and their revenues

top_two = descending_index[:2]

print("\nTop two products:")

for index in top_two:
    print(products[index], "€", revenues[index])
```
